梯度下降算法：
假设优化函数为：$$ y = f(x) = 0.5(x - 0.25)^{2} $$,即表示为损失函数

梯度的方向就代表这函数增长速度最快的方向，即x如何变化才会导致y值变化最快，（y值变化是按照y的曲线来变化的）如果要使得目标函数最小，梯度的反方向就是函数减小速度最快的方向，于是梯度下降公式为:$$ \theta=\theta-\alpha\frac{\partial J(\theta)}{\partial\theta} $$
其中$$\theta$$为待求参数，$$alpha$$为学习率表示每次优化的步长，$$J(\theta)$$表示目标函数。

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return 0.5*(x-0.25)**2
def df(x):
    return (x-0.25) #目标函数的导数
alpha = 0.1
x = -4
GD_X = [] #每次x更新后把值存在这个列表里面
GD_Y = [] #每次更新x后目标函数的值存在这个列表里面
x = -10 #随机初始化的x,其他的值也可以
f_current = f_change = f(x)
iter_num = 0
while iter_num <1000 and f_change > 1e-10: #迭代次数小于100次或者函数变化小于1e-10次方时停止迭代
    iter_num += 1
    alpha_t = alpha * df(x)
    x = x - alpha_t #梯度下降公式
    tmp = f(x)
    f_change = abs(f_current - tmp)
    f_current = tmp
    GD_X.append(x)
    GD_Y.append(f_current)


X = np.arange(-10,10,0.05)
Y = f(X)
Y = np.array(Y)

plt.plot(X,Y)
plt.scatter(GD_X,GD_Y)
plt.title("$y = 0.5(x - 0.25)^2$")
plt.show()
```

2.最小二乘法 error = 预测值 - 实际值，误差最小时就是最佳拟合关系

为避免误差之间正负的相互抵消，-->均方误差。

自变量不在是x而是参数w，因变量也变成了误差e。
3.代价函数如何使用：
一元二次函数，坐标系中为抛物线，机器学习的目标就是找到代价函数中的最小值。
